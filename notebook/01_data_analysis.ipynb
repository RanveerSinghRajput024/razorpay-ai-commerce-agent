```python
valid_invoices = invoice_product_count[
    (invoice_product_count >= 2) &
    (invoice_product_count <= 50)
].index
```

```python
len(valid_invoices)
```

```python
df_pairs = df_clean[
    df_clean["InvoiceNo"].isin(valid_invoices)
][["InvoiceNo", "StockCode", "Description"]].drop_duplicates()
```

```python
df_pairs.shape
```

```python
df_pairs.head(10)
```

```python
df_pairs["StockCode"] = df_pairs["StockCode"].astype(str)
```

```python
from itertools import combinations

product_pairs = []

for invoice, group in df_pairs.groupby("InvoiceNo"):
    products = group["StockCode"].unique()

    for product_a, product_b in combinations(sorted(products), 2):
        product_pairs.append((invoice, product_a, product_b))
```

```python
pairs_df = pd.DataFrame(
    product_pairs,
    columns=["InvoiceNo", "Product_A", "Product_B"]
)
pairs_df.shape
```

```python
pairs_df.head(20)
```

```python
pair_counts = (
    pairs_df
    .groupby(["Product_A", "Product_B"])
    .size()
    .reset_index(name="Purchase_Count")
)
```

```python
pair_counts = pair_counts.sort_values(
    "Purchase_Count",
    ascending=False
)
```

```python
pair_counts.head(20)
```

```python
product_order_count = (
    df_pairs
    .groupby("StockCode")["InvoiceNo"]
    .nunique()
    .reset_index(name="Order_Count")
)
product_order_count.head()
```

```python
product_order_count.shape
```

```python
# Create product pair counts
pair_counts = (
    pairs_df
    .groupby(["Product_A", "Product_B"])
    .size()
    .reset_index(name="Purchase_Count")
)

# Prepare product order counts
product_counts = product_order_count.rename(
    columns={
        "StockCode": "Product_A",
        "Order_Count": "Product_A_Order_Count"
    }
)

# Keep only the columns we need
product_counts = product_counts[
    ["Product_A", "Product_A_Order_Count"]
]

# Merge
pair_counts = pair_counts.merge(
    product_counts,
    on="Product_A",
    how="left"
)

# Calculate confidence
pair_counts["Confidence"] = (
    pair_counts["Purchase_Count"] /
    pair_counts["Product_A_Order_Count"]
)

# Sort by highest confidence
pair_counts = pair_counts.sort_values(
    "Confidence",
    ascending=False
)

# Show top 20
pair_counts.head(20)
```

```python
# Number of valid invoices used for our pair analysis
total_orders = len(valid_invoices)

# Calculate how many orders contain each Product B
product_b_counts = (
    df_pairs[df_pairs["InvoiceNo"].isin(valid_invoices)]
    .groupby("StockCode")["InvoiceNo"]
    .nunique()
    .reset_index(name="Product_B_Order_Count")
)

# Rename StockCode so we can merge it with Product_B
product_b_counts = product_b_counts.rename(
    columns={"StockCode": "Product_B"}
)

# Calculate support of Product B
product_b_counts["Support_B"] = (
    product_b_counts["Product_B_Order_Count"] / total_orders
)

# Merge Product B information
pair_counts = pair_counts.merge(
    product_b_counts[["Product_B", "Support_B"]],
    on="Product_B",
    how="left"
)

# Calculate Lift
pair_counts["Lift"] = (
    pair_counts["Confidence"] /
    pair_counts["Support_B"]
)

# Sort by highest Lift
pair_counts = pair_counts.sort_values(
    "Lift",
    ascending=False
)

# Show top 20
pair_counts.head(20)
```

```python
strong_pairs = pair_counts[
    pair_counts["Purchase_Count"] >= 20
].copy()
```

```python
strong_pairs = strong_pairs.sort_values(
    "Lift",
    ascending=False
)
```

```python
strong_pairs.head(20)
```

```python
# Strong practical recommendations
recommended_pairs = pair_counts[
    (pair_counts["Purchase_Count"] >= 100) &
    (pair_counts["Confidence"] >= 0.30)
].copy()

# Sort by confidence first, then purchase count
recommended_pairs = recommended_pairs.sort_values(
    ["Confidence", "Purchase_Count"],
    ascending=[False, False]
)

recommended_pairs.head(20)
```

```python
product_names = (
    df_clean[["StockCode", "Description"]]
    .dropna()
    .drop_duplicates("StockCode")
)

product_names["StockCode"] = product_names["StockCode"].astype(str)

recommended_pairs_named = recommended_pairs.merge(
    product_names.rename(
        columns={
            "StockCode": "Product_A",
            "Description": "Product_A_Name"
        }
    ),
    on="Product_A",
    how="left"
)

recommended_pairs_named = recommended_pairs_named.merge(
    product_names.rename(
        columns={
            "StockCode": "Product_B",
            "Description": "Product_B_Name"
        }
    ),
    on="Product_B",
    how="left"
)

recommended_pairs_named[
    [
        "Product_A",
        "Product_A_Name",
        "Product_B",
        "Product_B_Name",
        "Purchase_Count",
        "Confidence",
        "Lift"
    ]
].head(20)
```

```python
# Final recommendation dataset
final_recommendations = recommended_pairs_named[
    (recommended_pairs_named["Purchase_Count"] >= 100) &
    (recommended_pairs_named["Confidence"] >= 0.30)
].copy()

# Select only useful columns
final_recommendations = final_recommendations[
    [
        "Product_A",
        "Product_A_Name",
        "Product_B",
        "Product_B_Name",
        "Purchase_Count",
        "Confidence",
        "Lift"
    ]
]

# Sort by strongest practical recommendations
final_recommendations = final_recommendations.sort_values(
    ["Purchase_Count", "Confidence"],
    ascending=[False, False]
)

final_recommendations.head(20)
```

```python
import os

# Create data folder if it doesn't exist
os.makedirs("data", exist_ok=True)

# Save recommendation data
final_recommendations.to_csv(
    "data/product_relationships.csv",
    index=False
)

print("Saved successfully!")
print("Shape:", final_recommendations.shape)
```
